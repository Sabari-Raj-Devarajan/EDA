## Merging Employee and Salary Data



###DataFrames for Employees and Salaries

```python
import pandas as pd

df1_employees = pd.DataFrame({
    'EmployeeID': [1, 3, 5, 7],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'IT', 'Marketing', 'Sales']
})
df2_employees = pd.DataFrame({
    'EmployeeID': [2, 4, 6, 8, 9],
    'Name': ['Kevin', 'Liam', 'Mia', 'Noah', 'Olivia'],
    'Department': ['Sales', 'HR', 'IT', 'Marketing', 'Sales']
})

df1_salaries = pd.DataFrame({
    'EmployeeID': [1, 2, 5, 6, 10, 11],
    'Salary': [60000, 75000, 50000, 80000, 65000, 70000],
    'Bonus': [5000, 7500, 4000, 8000, 6000, 7000]
})
df2_salaries = pd.DataFrame({
    'EmployeeID': [3, 4, 7, 12, 13],
    'Salary': [68000, 82000, 53000, 79000, 67000],
    'Bonus': [6800, 8200, 5300, 7900, 6700]
})

print("Employee Data (df1_employees):\n")
display(df1_employees.head())
print("\nEmployee Data (df2_employees):\n")
display(df2_employees.head())
print("\nSalary Data (df1_salaries):\n")
display(df1_salaries.head())
print("\nSalary Data (df2_salaries):\n")
display(df2_salaries.head())
```

### Consolidate Employee and Salary DataFrames


```python
df_employees = pd.concat([df1_employees, df2_employees], ignore_index=True).drop_duplicates(subset=['EmployeeID'])
df_salaries = pd.concat([df1_salaries, df2_salaries], ignore_index=True).drop_duplicates(subset=['EmployeeID'])

print("Combined Employee Data (df_employees):\n")
display(df_employees.head())
print("\nCombined Salary Data (df_salaries):\n")
display(df_salaries.head())
```

### Merging Options for Employee and Salary Data


#### Option 1: Concatenate (axis=1)


```python
df_employees_indexed = df_employees.reset_index(drop=True)
df_salaries_indexed = df_salaries.reset_index(drop=True)

df_concat_axis1 = pd.concat([df_employees_indexed, df_salaries_indexed], axis=1)
print("Concatenated Data (axis=1):\n")
display(df_concat_axis1.head())
```

#### Option 2: Inner Merge

```python
df_inner_merge = df_employees.merge(df_salaries, on='EmployeeID', how='inner')
print("Inner Merge (Employees with both info and salary):\n")
display(df_inner_merge.head())
print(f"Number of employees with both records: {len(df_inner_merge)}")
```

#### Option 3: Left Merge

```python
df_left_merge = df_employees.merge(df_salaries, on='EmployeeID', how='left')
print("Left Merge (All employees, with salary info if available):\n")
display(df_left_merge.head())
print(f"Total employees in the left merge: {len(df_left_merge)}")
```

#### Option 4: Right Merge


```python
df_right_merge = df_employees.merge(df_salaries, on='EmployeeID', how='right')
print("Right Merge (All salary records, with employee info if available):\n")
display(df_right_merge.head())
print(f"Total salary records in the right merge: {len(df_right_merge)}")

missing_employee_info = df_right_merge[df_right_merge['Name'].isnull()]
num_missing_employee_info = len(missing_employee_info)
print(f"\nNumber of salary records with no matching employee info: {num_missing_employee_info}")

if num_missing_employee_info > 0:
    print("Salary records with no matching employee details:\n")
    display(missing_employee_info)
```

## Reshaping with Hierarchical Indexing for Employee Data


```python
df_hierarchical = df_inner_merge.set_index(['Department', 'Name'])
print("DataFrame with Hierarchical Index:\n")
display(df_hierarchical)
```

```python
stacked_data = df_hierarchical.stack()
print("Stacked Data:\n")
display(stacked_data)
```

```python
unstacked_data = stacked_data.unstack()
print("Unstacked Data (default):\n")
display(unstacked_data)
```

```python
unstacked_by_dept = df_hierarchical.unstack(level='Department')
print("Unstacked by Department:\n")
display(unstacked_by_dept)
```
