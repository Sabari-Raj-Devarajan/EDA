```python
# 1. Armstrong Number
n = int(input("Enter a number: "))
temp = n
digits = len(str(n))
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** digits
    temp //= 10

if sum == n:
    print("Armstrong Number")
else:
    print("Not Armstrong")
#25BAI0158
```

```python
# 2. Factorial
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
#25BAI0158
```

```python
# 3. Prime Number Check
n = int(input("Enter a number: "))
if n > 1:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
#25BAI0158
```

```python
# 4. LCM Calculation
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

greater = max(a, b)

while True:
    if greater % a == 0 and greater % b == 0:
        print("LCM =", greater)
        break
    greater += 1
#25BAI0158
```

```python
# 5. Palindrome Check
s = input("Enter a string: ")
rev = s[::-1]
if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
#25BAI0158
```

```python
# 6. Even and Odd Count
arr = list(map(int, input().split()))

even = 0
odd = 0

for i in arr:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even =", even)
print("Odd =", odd)
#25BAI0158
```

```python
# 7. Matrix Multiplication
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

C = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            C[i][j] += A[i][k] * B[k][j]

print(C)
#25BAI0158
```

```python
# 8. Swapping Two Numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Before swapping: a =", a, ", b =", b)
a, b = b, a
print("After swapping: a =", a, ", b =", b)
print(a, b)
#25BAI0158
```

```python
# 9. Bubble Sort
arr = list(map(int, input("Enter elements: ").split()))

n = len(arr)

for i in range(n):
    for j in range(0, n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

print("Sorted Array:", arr)
#25BAI0158
```

```python
# 10. Character Frequency Counter
s = input()

visited = ""

for ch in s:
    if ch not in visited:
        count = 0
        for x in s:
            if ch == x:
                count += 1
        print(ch, "=", count)
        visited += ch
#25BAI0158
```
