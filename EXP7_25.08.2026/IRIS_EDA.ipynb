```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset
iris_df = pd.read_csv('/content/Iris.csv')
```

### 1. Data Overview

```python
print("First 5 rows of the Iris dataset:")
display(iris_df.head())
```

```python
print("\nDataFrame Information:")
iris_df.info()
```

```python
print("\nDescriptive Statistics:")
display(iris_df.describe())
```

```python

print("\nDistribution of Species:")
display(iris_df['Species'].value_counts())
```

### 2. Visualizing Data Distributions

```python
iris_df.hist(figsize=(10, 8))
plt.suptitle('Histograms of Iris Features', y=1.02)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))
for i, feature in enumerate(iris_df.columns[1:-1]):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(x='Species', y=feature, data=iris_df)
    plt.title(f'{feature} by Species')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))
# Exclude 'Id' column (index 0) and 'Species' column (last)
for i, feature in enumerate(iris_df.columns[1:-1]):
    plt.subplot(2, 2, i + 1)
    sns.violinplot(x='Species', y=feature, data=iris_df)
    plt.title(f'{feature} by Species')
plt.tight_layout()
plt.show()
```

### 3. Relationships Between Features


```python
sns.pairplot(iris_df, hue='Species', markers=["o", "s", "D"])
plt.suptitle('Pair Plot of Iris Features by Species', y=1.02)
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap
plt.show()
```
